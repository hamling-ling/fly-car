# 条件を変えて走る

基準の 1 場面は `notebooks/fov_sim.ipynb` にある。ここでは球の位置だけを変える。カメラは対角 58°、高さ 0.15 m、俯角 12° のまま。車は原点から +x を向いて出る。

数を変えたくなったら、節をコピーして `Scene` を書き換える。


```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "flycar" / "behavior.py").exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from flycar.fov import Blob, Scene, simulate
from flycar.view import neuron_figure, show_world
```

## 何もない

画角に青も赤もないときは動かない。

```python
scene = Scene(
    red=None,
    blue=None,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 正面の赤


赤を `(1.30, 0)` m、半径 0.12 m に置く。青は置かない。

向きは 0 のまま直進し、6.46 秒に止まる。両輪は同じ速さである。曲がらない分、止まるのは基準場面より早い。


```python
scene = Scene(
    red=Blob((1.30, 0.00), 0.12),
    blue=None,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 左の赤


赤を `(1.30, +0.40)` m、半径 0.12 m に置く。青は置かない。基準場面の左右を入れ替えたものである。

左へ曲がり、0.26 秒に中央へ入る。一度中央を越えてから戻り、6.86 秒に止まる。


```python
scene = Scene(
    red=Blob((1.30, 0.40), 0.12),
    blue=None,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 右の青だけ


赤は置かない。青を `(0.45, -0.08)` m、半径 0.10 m に置く。開始時は右の半画面が 0.83、左が 0.09 で、回避の閾値 0.72 は右だけが超える。

0.22 秒まで、その場で左へ回る。左輪は後ろ、右輪は前である。赤が無いので、青が 0.72 を下回ったあとも旋回の指令は青のままである。曲がりながら進み、0.82 秒に青が 0.15 を下回って止まる。位置はおよそ (0.09, 0.03) m、向きは左へ約 20° である。


```python
scene = Scene(
    red=None,
    blue=Blob((0.45, -0.08), 0.10),
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 正面の赤と右の青


正面の赤と、上と同じ青を同時に置く。青が右に映っているあいだは、赤の向きを使わず、前進したまま左へ避ける。

向きは 0.38 秒で約 11° まで増える。青のカウントは閾値の上下を行き来するので、舵がそのあいだ入れ替わる。青が画面から出たあと赤へ戻り、6.60 秒に赤の前で止まる。


```python
scene = Scene(
    red=Blob((1.30, 0.00), 0.12),
    blue=Blob((0.45, -0.08), 0.10),
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```
