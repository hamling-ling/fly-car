# 条件を変えて走る

基準の 1 場面は `notebooks/fov_sim.ipynb` にある。ここでは球の位置だけを変える。カメラは対角 58°、高さ 0.15 m、俯角 12° のまま。車は原点から +x を向いて出る。

数を変えたくなったら、節をコピーして `Scene` を書き換える。


```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "src" / "flycar" / "behavior.py").exists():
    ROOT = ROOT.parent
SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.insert(0, str(SRC))

from flycar.fov import Blob, Scene, simulate
from flycar.view import neuron_figure, show_world
```

## 何もない

画角に青も赤もないときは動かない。

```python
scene = Scene(
    red=None,
    blue=None,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 正面の赤


赤を `(1.30, 0)` m、半径 0.12 m に置く。青は置かない。

向きは 0 のまま直進し、6.44 秒に止まる。両輪は同じ速さである。赤が近づくほど発火が増えて速くなる。曲がらない分、止まるのは基準場面より早い。


```python
scene = Scene(
    red=Blob((1.30, 0.00), 0.12),
    blue=None,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 左の赤


赤を `(1.30, +0.40)` m、半径 0.12 m に置く。青は置かない。基準場面の左右を入れ替えたものである。

左へ曲がり、3.74 秒に中央へ入る。向きは左へ約 20° まで回って赤の方向を向いたまま近づき、7.88 秒に止まる。


```python
scene = Scene(
    red=Blob((1.30, 0.40), 0.12),
    blue=None,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 右の青だけ


赤は置かない。青を `(0.45, -0.08)` m、半径 0.10 m に置く。開始時は右の半画面が 0.83、左が 0.09 である。

前進は赤の和から青の和の一部を引いたものなので、赤が無いと前進が残らない。その場で左へ回る。青が右端へ寄って半画面の占有が床を割ると発火が消え、1.76 秒に止まる。位置は原点のまま、向きは左へ約 26° である。右の半画面の青は約 0.04 まで下がっている。


```python
scene = Scene(
    red=None,
    blue=Blob((0.45, -0.08), 0.10),
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 正面の赤と右の青


正面の赤と、上と同じ青を同時に置く。青のヨーを赤より強くしてあるので、右の青を左に避けながら進む。向きはいったん左へ約 24° まで回り、青は 2.3 秒ごろ画面から出る。最接近は 5.40 秒で、青の中心から 0.18 m、半径 0.10 m の外側である。そのあと赤の方へ戻り、8.88 秒に赤の前で止まる。この節だけ走行を 12 秒にして、停止まで入れてある。両輪は前進のままである。


```python
scene = Scene(
    red=Blob((1.30, 0.00), 0.12),
    blue=Blob((0.45, -0.08), 0.10),
    run_s=12.0,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```
