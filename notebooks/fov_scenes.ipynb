# 条件を変えて走る

基準の 1 場面は `notebooks/fov_sim.ipynb` にある。ここでは球の位置だけを変える。カメラは対角 58°、高さ 0.15 m、俯角 12° のまま。車は原点から +x を向いて出る。

数を変えたくなったら、節をコピーして `Scene` を書き換える。


```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "flycar" / "behavior.py").exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from flycar.fov import Blob, Scene, simulate
from flycar.view import neuron_figure, show_world
```

## 何もない

画角に青も赤もないときは動かない。

```python
scene = Scene(
    red=None,
    blue=None,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 正面の赤


赤を `(1.30, 0)` m、半径 0.12 m に置く。青は置かない。

向きは 0 のまま直進し、6.44 秒に止まる。両輪は同じ速さである。赤が近づくほど発火が増えて速くなる。曲がらない分、止まるのは基準場面より早い。


```python
scene = Scene(
    red=Blob((1.30, 0.00), 0.12),
    blue=None,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 左の赤


赤を `(1.30, +0.40)` m、半径 0.12 m に置く。青は置かない。基準場面の左右を入れ替えたものである。

左へ曲がり、0.94 秒に中央へ入る。1.1 秒にわずかに右へ越えて、1.6 秒には戻る。7.22 秒に止まる。


```python
scene = Scene(
    red=Blob((1.30, 0.40), 0.12),
    blue=None,
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 右の青だけ


赤は置かない。青を `(0.45, -0.08)` m、半径 0.10 m に置く。開始時は右の半画面が 0.83、左が 0.09 である。

前進は赤の和から青の和を引いたものなので、赤が無いと前進が残らない。その場で左へ回る。青が画面の端へ寄って電流が閾値を割ると発火が消え、2.02 秒に止まる。位置は原点のまま、向きは左へ約 13° である。右の半画面には青が約 0.59 残る。


```python
scene = Scene(
    red=None,
    blue=Blob((0.45, -0.08), 0.10),
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```

## 正面の赤と右の青


正面の赤と、上と同じ青を同時に置く。少し左を向くと赤も右の半画面に入り、右の DNa02 と右の DNp01 がほぼ同じ率で逆向きに出る。和が相殺されるのでほとんど進まず、8 秒で約 3 cm、向きは左へ約 1° である。赤の前には着かない。車輪がゼロ付近を行き来するので、図のモードは前進とその場旋回が細かく入れ替わる。


```python
scene = Scene(
    red=Blob((1.30, 0.00), 0.12),
    blue=Blob((0.45, -0.08), 0.10),
)
run = simulate(scene)
print(run)
```

```python
show_world(run)
```

```python
neuron_figure(run)
```
